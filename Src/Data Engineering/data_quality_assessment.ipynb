```python

def basic_data_quality(df, name):

    print("\n" + "=" * 60)
    print(f"DATA QUALITY REPORT: {name}")
    print("=" * 60)

    # 1. Dataset dimensions
    print("\n1. Dataset Shape")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    # 2. Data types
    print("\n2. Data Types")
    print(df.dtypes)

    # 3. Missing values
    missing = df.isna().sum()

    # 4. Empty strings
    empty = (
        df.astype("string")
        .apply(lambda col: col.str.strip().eq("").sum())
    )

    print("\n3. Missing and Empty Values")

    report = pd.DataFrame({
        "Missing": missing,
        "Empty Strings": empty
    })

    print(
        report[
            (report["Missing"] > 0) |
            (report["Empty Strings"] > 0)
        ]
    )

    # 5. Duplicate records
    print("\n4. Duplicate Rows")
    print("Duplicates:", df.duplicated().sum())

    # 6. Unique values
    print("\n5. Unique Values")
    print(df.nunique())
```

```python

for name, df in datasets.items():

    basic_data_quality(df, name)
```

```python

def analyze_empty_values(df, name):

    print("\n" + "=" * 60)
    print(f"EMPTY VALUES ANALYSIS: {name}")
    print("=" * 60)

    # Count empty strings in each column
    empty_counts = (
        df.astype("string")
        .apply(lambda col: col.str.strip().eq("").sum())
    )

    # Calculate percentage
    empty_percentage = (
        empty_counts / len(df) * 100
    )

    # Create report
    report = pd.DataFrame({
        "Empty Count": empty_counts,
        "Empty Percentage": empty_percentage.round(2)
    })

    # Show only columns containing empty values
    report = report[report["Empty Count"] > 0]

    # Sort by empty count
    report = report.sort_values(
        by="Empty Count",
        ascending=False
    )

    display(report)

    return report


regional_empty_report = analyze_empty_values(
    df_regional, "Regional"
)
```

```python

# Create a temporary copy
national_time_analysis = df_national.copy()

# Convert date column to datetime
national_time_analysis["parsed_date"] = pd.to_datetime(
    national_time_analysis["data"],
    errors="coerce"
)

# Extract year
national_time_analysis["year"] = (
    national_time_analysis["parsed_date"].dt.year
)

columns_to_check = [
    "note_casi",
    "note_test",
    "note",
    "tamponi_test_molecolare",
    "totale_positivi_test_molecolare",
    "tamponi_test_antigenico_rapido",
    "ingressi_terapia_intensiva"
]

for col in columns_to_check:

    print("\n" + "=" * 60)
    print("YEARLY MISSING ANALYSIS:", col)
    print("=" * 60)

    is_empty = (
        national_time_analysis[col]
        .astype("string")
        .str.strip()
        .eq("")
    )

    yearly_report = (
        national_time_analysis
        .assign(is_empty=is_empty)
        .groupby("year")
        .agg(
            Total_Rows=("is_empty", "size"),
            Empty_Values=("is_empty", "sum")
        )
    )

    yearly_report["Available_Values"] = (
        yearly_report["Total_Rows"] -
        yearly_report["Empty_Values"]
    )

    yearly_report["Missing_Percentage"] = (
        yearly_report["Empty_Values"] /
        yearly_report["Total_Rows"] * 100
    ).round(2)

    display(yearly_report)
```

```python

# Verify Missingness Patterns Across Regions

availability_comparison = []

for col in regional_columns:

    temp = regional_df[
        ["parsed_date", "codice_regione", col]
    ].copy()

    temp["is_available"] = (
        temp[col].notna() &
        temp[col].astype("string").str.strip().ne("")
    )

    availability_pivot = temp.pivot(
        index="parsed_date",
        columns="codice_regione",
        values="is_available"
    )

    values_pivot = temp.pivot(
        index="parsed_date",
        columns="codice_regione",
        values=col
    )

    # Compare missingness patterns
    same_missing_pattern = (
        availability_pivot.nunique(axis=1).le(1).all()
    )

    # Compare actual values across regions
    common_dates = values_pivot.dropna()

    same_actual_values = (
        common_dates.nunique(axis=1).le(1).all()
        if not common_dates.empty else None
    )

    availability_comparison.append({
        "Column": col,
        "Same_Missing_Pattern": same_missing_pattern,
        "Same_Actual_Values": same_actual_values,
        "Dates_Compared": len(availability_pivot),
        "Dates_With_Different_Values": (
            common_dates.nunique(axis=1).gt(1).sum()
        )
    })

availability_comparison_df = pd.DataFrame(
    availability_comparison
)

display(availability_comparison_df)
```
